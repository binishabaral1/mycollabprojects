```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
X_test.shape
```

```python
y_train
```

```python
import matplotlib.pyplot as plt
plt.imshow(X_train[3])
```

```python
X_train = X_train/255
X_test = X_test/255
```

```python
X_train[3]
```

```python
model = Sequential()
model.add(Flatten(input_shape=(28,28)))
model.add(Dense(128, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='Adam', metrics=['accuracy'])
```

```python
history = model.fit(X_train, y_train, epochs=10, validation_split=0.2)
```

```python
y_prob = model.predict(X_test)
```

```python
y_pred = y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
from PIL import Image
import numpy as np

# Load the image
img = Image.open('ssavr.jpeg')

# Convert to grayscale and resize to 28x28
img = img.convert('L') # 'L' for grayscale
img = img.resize((28, 28))

# Convert to numpy array and normalize
img_array = np.array(img)
img_array = img_array / 255.0

# Reshape for model input (add batch dimension)
img_for_prediction = np.expand_dims(img_array, axis=0)

print("Shape of processed image:", img_for_prediction.shape)

# You can now use img_for_prediction with your model
# For example: prediction = model.predict(img_for_prediction)
```

```python
prediction_prob = model.predict(img_for_prediction)
predicted_digit = prediction_prob.argmax(axis=1)[0]
print(f"The model predicts the digit: {predicted_digit}")
```
