Step 8:Chat Word Treatment

```python
import requests

url = "https://raw.githubusercontent.com/rishabhverma17/sms_slang_translator/master/slang.txt"
response = requests.get(url)

with open("chat_words.txt", "w") as f:
    f.write(response.text)
```

```python
chat_words = {}

with open("chat_words.txt", "r") as f:
    for line in f:
        if "=" in line:
            key, value = line.strip().split("=")
            chat_words[key.lower()] = value.lower()

chat_words
```

```python
def chat_word_treatment(text):
    words = text.split()
    return " ".join(chat_words.get(word, word) for word in words)
```

```python
df['chatword_test'] = df['review'].apply(chat_word_treatment)
df[['review', 'chatword_test']].head()
```

Remove Punctuation

```python
def remove_punctuation(text):
    return re.sub(r'[^\w\s]', '', text)
```

```python
remove_punctuation(df['review'][0])
```

```python
def spelling_correction(text):
    return str(TextBlob(text).correct())
```

Remove stop words

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    return " ".join(word for word in words if word not in stop_words)
```

```python
remove_stopwords(df['review'][0])
```

Tokenization

```python
from nltk.tokenize import word_tokenize

def tokenize(text):
    return word_tokenize(text)
```
